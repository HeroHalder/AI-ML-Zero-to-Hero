```python
monthes=["January","February","March","April"]
Views24=[1000,2000,3000,4000]
Views25=[2000,3000,4000,5000]
Views26=[3000,4000,5000,6000]
```

```python
#Multi data in one figure
plt.figure()
plt.plot(monthes,Views24)
plt.plot(monthes,Views25)
plt.plot(monthes,Views26)
```

```python
plt.plot(monthes,Views24,label="2024 views")
plt.plot(monthes,Views25,label="2025 views")
plt.plot(monthes,Views26,label="2026 views")
plt.legend(loc="best") #upper left
```

```python
x=np.arange(0,12)
y=x**2
z=x**3
a=x
```

```python
fig,axes=plt.subplots(2,2)
axes[0,0].plot(x,y)
axes[0,1].scatter(x,a)
axes[1][0].plot(x,z)
axes[1,1].scatter(x,z)
```

```python
fig,axes=plt.subplots(2,2)
axes[0,0].plot(x,y)
axes[0,1].scatter(x,a)
axes[1][0].plot(x,z)
axes[1,1].scatter(x,z)
fig.suptitle("My Graph")
fig.tight_layout()
fig.suptitle("My Graph")
```

```python
fig,axes=plt.subplots(2,2)
axes[0,0].plot(x,y)
axes[0,0].set_title("axis 0,0")
axes[0,1].scatter(x,a)
axes[1][0].plot(x,z)
axes[1,1].scatter(x,z)
fig.suptitle("My Graph")
fig.tight_layout()
fig.suptitle("My Graph")
fig.subplots_adjust(wspace=1,hspace=1)
```
